```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage,SystemMessage,ToolMessage
from langchain_core.prompts import HumanMessagePromptTemplate,SystemMessagePromptTemplate,ChatPromptTemplate
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import SQLChatMessageHistory
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv(override=True)
parser = StrOutputParser()
llm = ChatOpenAI(model="gpt-5.4-mini")
# sqlite:///chat_history.db
```

```python
def get_chat_history(session_id:str):
    return SQLChatMessageHistory(session_id=session_id,connection="sqlite:///chat_history.db")
```

```python
user_id = "test123"
```

```python
template = ChatPromptTemplate.from_template("{prompt}")
chain = template | llm | parser
runable_with_history = RunnableWithMessageHistory(chain,get_chat_history)
result = runable_with_history.invoke([HumanMessage(content="Hi, I'm Ravi and I want to visit Goa")],config={"configurable":{"session_id":user_id}})
print(result)
```
