```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage,SystemMessage,ToolMessage
from langchain_core.prompts import SystemMessagePromptTemplate,ChatPromptTemplate,HumanMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import SQLChatMessageHistory
```

```python
parser = StrOutputParser()
llm  = ChatOpenAI(model="gpt-5.4-mini")

template = ChatPromptTemplate.from_template("{prompt}")
chain = template | llm |parser
about = "My name is ganesh.I work for discovery"
output = chain.invoke({"prompt":about})
print(output)
```

```python
prompt = "what is my name?"
output = chain.invoke({"prompt":prompt})
print(output)
```

```python
from langchain_community.chat_message_histories import SQLChatMessageHistory

def get_session_history(session_id):
    return SQLChatMessageHistory(
        session_id=session_id,
        connection="sqlite:///chat_history.db",
    )
```

```python
runnable_with_hisory = RunnableWithMessageHistory(chain,get_session_history)
```

```python
user_id = 'ganesh1234'
history = get_session_history(user_id)
print(history.get_messages())
```
