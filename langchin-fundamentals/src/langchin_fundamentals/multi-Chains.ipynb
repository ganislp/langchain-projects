```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate,ChatPromptTemplate
from langchain_core.messages import SystemMessage,HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv(override=True)

llm = ChatOpenAI(model="gpt-5.4-mini")
llm_groq = ChatGroq(temperature=0,model="openai/gpt-oss-120b",api_key=os.getenv("GROQ_API_KEY"))

parser = StrOutputParser()
```

```python
system_prompt = SystemMessagePromptTemplate.from_template("You are a chef. The user is {diet}. Never use {avoid}")
user_prompt = HumanMessagePromptTemplate.from_template("Give me a recipe for {dish}")

messages = [system_prompt,user_prompt]

template = ChatPromptTemplate(messages)
chain = template | llm | parser
# result = chain.invoke({"diet":"vegan","avoid":"nuts","dish":"Banoffee Pie"})
# print(result)
```

```python
from operator import itemgetter
# analysis_prompt = ChatPromptTemplate.from_template(''' analyze the following text: {text}
# you need tell me that how diffuclt it is to understand Anawer in one sentence only''')
converter_prompt =  ChatPromptTemplate.from_template("Translate the following text: {text}\ninto {target} language")
converter_chain = converter_prompt | llm_groq | parser

# output = converter_chain.invoke({'text':result,"target":"telugu"})
# print(output)
composed_chain = (
    {"text":chain,"target": itemgetter("target")}
    | converter_prompt
    | llm_groq
    | parser
)

output = composed_chain.invoke({
    "diet": "vegan",
    "avoid": "nuts",
    "dish": "Banoffee Pie",
    "target": "telugu",
})
print(output)
```
