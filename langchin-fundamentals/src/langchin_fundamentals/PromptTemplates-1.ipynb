```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage,HumanMessage,ToolMessage
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate,PromptTemplate,ChatPromptTemplate
```

```python
load_dotenv(override=True)
```

```python
llm = ChatOpenAI(model="gpt-5.4-mini")
# system_message = SystemMessage("You are a english teacher for beginner students. Correct mistakes politely.")
# user_message = HumanMessage("She don’t like the coffee")
# messages = [system_message,user_message]
# result = llm.invoke(messages);
# print(result.content)
```

```python
system_prompt = SystemMessagePromptTemplate.from_template("You are a {language} teacher for {level} students. Correct mistakes politely.")
user_prompt = HumanMessagePromptTemplate.from_template("{sentence}")

messages = [system_prompt,user_prompt]

template = ChatPromptTemplate(messages)

quection = template.invoke({"language":"english","level":"beginner","sentence":"Me and my friend plays cricket on Sunday."})

result = llm.invoke(quection)
print(result.content)
```

```python
system_prompt = SystemMessagePromptTemplate.from_template("You are a chef. The user is {diet}. Never use {avoid}")
user_prompt = HumanMessagePromptTemplate.from_template("Give me a recipe for {dish}")

messages = [system_prompt,user_prompt]

template = ChatPromptTemplate(messages)

quection = template.invoke({"diet":"vegan","avoid":"nuts","dish":"Banoffee Pie"})

result = llm.invoke(quection)
print(result.content)
```

```python
system_prompt = SystemMessagePromptTemplate.from_template("You are a chef. The user is {diet}. Never use {avoid}")
user_prompt = HumanMessagePromptTemplate.from_template("Give me a recipe for {dish}")
messages = [system_prompt,user_prompt]
template = ChatPromptTemplate(messages)
chain = template | llm
result = chain.invoke({"diet":"vegan","avoid":"nuts","dish":"Banoffee Pie"})
print(result.content)
```
