```python
load_dotenv(override=True)
llm = ChatOpenAI(model="gpt-5.4-mini")
```

```python
system_prompt = SystemMessagePromptTemplate.from_template("you are a help full {language} developer")
user_prompt = HumanMessagePromptTemplate.from_template("tell me about {language} defuculat program loop 1 to millon simple way no need explanation")
```

```python
# user_prompt.format(language='python')
# system_prompt.format(language='python')

messages = [system_prompt,user_prompt]

template = ChatPromptTemplate(messages)
question = template.invoke({"language":"java"})

result = llm.invoke(question)
print(result.content)
```
