```python
import os
import warnings
warnings.filterwarnings("ignore")
```

```python
from markitdown import MarkItDown
from pathlib import Path
```

```python
md = MarkItDown()

file_path = Path("../../rag-dataset/data/Apple-10-Q-2025-Q1.pdf")
file_path.stem
```

```python
base_path = Path("../../rag-dataset/data")
```

```python
result = md.convert(file_path)
os.makedirs(base_path/"markitdown",exist_ok=True)

with open(base_path / f"markitdown/{file_path.stem}.md","w",encoding="utf-8") as f:
     f.write(result.text_content)
```

```python
def convert_and_save(file_path):
    file_path = Path(file_path)
    result = md.convert(file_path) 
    with open(base_path / f"markitdown/{file_path.stem}.md","w",encoding="utf-8") as f:
       f.write(result.text_content)
```
