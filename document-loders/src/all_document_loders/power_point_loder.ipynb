```python
import os
from langchain_community.document_loaders import UnstructuredPowerPointLoader
from pathlib import Path
from scripts import llm
```

```python
base_path = Path("../../rag-dataset/data")
pptx_path = base_path / "ml_course.pptx"

loader = UnstructuredPowerPointLoader(str(pptx_path),mode="elements")
docs = loader.load()
len(docs)

```

```python
docs
```

```python
doc = docs[0]
# doc.metadata
# doc.page_content
```

```python
ppt_data = {}
for doc in docs:
    page = doc.metadata["page_number"]
    ppt_data[page] = ppt_data.get(page,"") + "\n\n" +doc.page_content
```

```python
ppt_data
```

```python
context = ""
for page,content in ppt_data.items():
    context += f"### Slide {page} \n\n{content.strip()}\n\n\n"
```

```python
question ="""
For each PowerPoint slide provided above, write a 2-minute script that effectively conveys the key points.
Ensure a smooth flow between slides, maintaining a clear and engaging narrative.
"""
```

```python
response = llm.call_llm(context,question)
with open(base_path / "ml_course.md","w") as f:
     f.write(response)
```
