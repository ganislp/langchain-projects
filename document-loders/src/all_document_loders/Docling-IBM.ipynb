```python
from docling.document_converter import DocumentConverter
import os
from pathlib import Path
```

```python
converter = DocumentConverter()
```

```python
file_path = Path("../../rag-dataset/data/Apple-10-Q-2025-Q1.pdf")
file_path.stem
```

```python
result = converter.convert(file_path)
```

```python
base_path = Path("../../rag-dataset/data")
markdown = result.document.export_to_markdown()
os.makedirs(base_path/"docling",exist_ok=True)

with open(base_path / f"docling/{file_path.stem}.md","w",encoding="utf-8") as f:
     f.write(markdown)
```

```python
def convert_and_save(file_path):
    file_path = Path(file_path)
    result = converter.convert(file_path)
    markdown = result.document.export_to_markdown()
    with open(base_path / f"docling/{file_path.stem}.md","w",encoding="utf-8") as f:
     f.write(markdown)
```

```python
convert_and_save("../../rag-dataset/data/scansmpl.pdf")
```
